```python
import numpy as np
import healpy as hp

power_spectrum_file = "COM_PowerSpect_CMB-TT-full_R3.01.txt"

output_file = (
    "/home/agnes/Skyclean/docs/circle/"
    "gaussian_null_lmax800_nside512.fits"
)

nside = 512
lmax = 800
seed = 12345

# Read D_ell [microK^2]
data = np.loadtxt(power_spectrum_file)

ells = data[:, 0].astype(int)
Dl_microK2 = data[:, 1]

# Convert D_ell [microK^2] -> C_ell [K^2]
cl = np.zeros(lmax + 1)

for ell, Dl in zip(ells, Dl_microK2):
    if 2 <= ell <= lmax:
        cl[ell] = (
            Dl * 1e-12
            * 2.0 * np.pi
            / (ell * (ell + 1.0))
        )

# Generate one full-sky Gaussian CMB map
np.random.seed(seed)

alm = hp.synalm(cl, lmax=lmax, new=True)

cmb_map = hp.alm2map(
    alm,
    nside=nside,
    lmax=lmax,
    verbose=False,
)

hp.write_map(
    output_file,
    cmb_map,
    overwrite=True,
)

print("Saved:", output_file)
print("NSIDE:", hp.get_nside(cmb_map))
print("Mean:", np.mean(cmb_map))
print("Std:", np.std(cmb_map))
```

```python
import glob
import numpy as np
import healpy as hp
import matplotlib.pyplot as plt

# -------------------------
# Files
# -------------------------
true_map = hp.read_map("ilc_null_64_nd.fit", field=0, verbose=False)

cmbpaint_files = sorted(glob.glob("inpainted_maps/Inpainted_Maps/Inpainted_T_*.fits"))[:100]
iter_files = sorted(glob.glob("iterative_inpainted_maps/*.fits"))[:100]

print("CMB-PAInT files:", len(cmbpaint_files))
print("Iterative files:", len(iter_files))

# -------------------------
# Settings
# -------------------------
nside = hp.get_nside(true_map)
192
ells = np.arange(lmax + 1)

true_map = hp.remove_dipole(true_map, verbose=False)
true_map = hp.remove_monopole(true_map, verbose=False)
cl_true = hp.anafast(true_map, lmax=lmax)

# -------------------------
# Mean power spectrum of each method
# -------------------------
def mean_cl(files):
    cls = []

    for f in files:
        m = hp.read_map(f, field=0, verbose=False)

        if hp.get_nside(m) != nside:
            m = hp.ud_grade(m, nside_out=nside)

        m = hp.remove_dipole(m, verbose=False)
        m = hp.remove_monopole(m, verbose=False)

        cl = hp.anafast(m, lmax=lmax)
        cls.append(cl)

    return np.mean(cls, axis=0)

cl_cmbpaint = mean_cl(cmbpaint_files)
cl_iter = mean_cl(iter_files)

# -------------------------
# Difference relative to input null
# -------------------------
delta_cmbpaint = cl_cmbpaint - cl_true
delta_iter = cl_iter - cl_true

# -------------------------
# Variance over ell bands
# -------------------------
low = (ells >= 2) & (ells <= 20)
high = (ells >= 21) & (ells <= 94)

print("Variance CMB-PAInT - input null, ell 2-20:")
print(np.var(delta_cmbpaint[low]))

print("Variance Iterative - input null, ell 2-20:")
print(np.var(delta_iter[low]))

print("\nVariance CMB-PAInT - input null, ell 21-94:")
print(np.var(delta_cmbpaint[high]))

print("Variance Iterative - input null, ell 21-94:")
print(np.var(delta_iter[high]))

# -------------------------
# Plot
# -------------------------
plt.figure(figsize=(7, 5))

plt.plot(
    ells[2:],
    delta_cmbpaint[2:],
    marker="o",
    label=r"$C_\ell^{\rm CMB\text{-}PAInT} - C_\ell^{\rm input}$"
)

plt.plot(
    ells[2:],
    delta_iter[2:],
    marker="o",
    label=r"$C_\ell^{\rm iterative} - C_\ell^{\rm input}$"
)

# plt.xlim(40, 93)
# plt.ylim(-1e-11, 1e-11)
plt.axhline(0.0, linestyle="--", linewidth=1, color="black")

plt.xlabel(r"$\ell$")
plt.ylabel(r"$\Delta C_\ell$")
plt.title("Power-spectrum residual relative to input null simulation")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
import glob
import re
import healpy as hp
import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# Files
# -------------------------
null_file = "ilc_null_64_nd.fit"

def natural_key(path):
    return [int(t) if t.isdigit() else t for t in re.split(r"(\d+)", path)]

cmbpaint_files = sorted(
    glob.glob("inpainted_maps/Inpainted_Maps/Inpainted_T_*.fits"),
    key=natural_key
)

iter_files = sorted(
    glob.glob("iterative_inpainted_maps/*.fits"),
    key=natural_key
)

print("Number of CMB-PAInT files:", len(cmbpaint_files))
print("Number of iterative files:", len(iter_files))

# -------------------------
# Load null map
# -------------------------
null_map = hp.read_map(null_file, field=0, verbose=False)
nside = hp.get_nside(null_map)

# -------------------------
# Load and average realizations
# -------------------------
def load_mean_map(files, nside):
    maps = []

    for f in files:
        m = hp.read_map(f, field=0, verbose=False)

        if hp.get_nside(m) != nside:
            m = hp.ud_grade(m, nside_out=nside)

        maps.append(m)

    return np.mean(np.stack(maps, axis=0), axis=0)

cmbpaint_map = load_mean_map(cmbpaint_files, nside)
iter_map = load_mean_map(iter_files, nside)

# -------------------------
# Remove bad pixels
# -------------------------
good = (
    np.isfinite(null_map)
    & np.isfinite(cmbpaint_map)
    & np.isfinite(iter_map)
    & (null_map != hp.UNSEEN)
    & (cmbpaint_map != hp.UNSEEN)
    & (iter_map != hp.UNSEEN)
)

null_map = null_map[good]
cmbpaint_map = cmbpaint_map[good]
iter_map = iter_map[good]

# -------------------------
# Common bins
# -------------------------
all_vals = np.concatenate([null_map, cmbpaint_map, iter_map])
xmin = np.percentile(all_vals, 0.1)
xmax = np.percentile(all_vals, 99.9)
bins = np.linspace(xmin, xmax, 150)

# -------------------------
# Plot
# -------------------------
plt.figure(figsize=(8, 5))

plt.hist(
    null_map,
    bins=bins,
    density=True,
    histtype="step",
    linewidth=2,
    label="Null simulation"
)

plt.hist(
    cmbpaint_map,
    bins=bins,
    density=True,
    histtype="step",
    linewidth=2,
    label="CMB-PAInT ensemble mean"
)

plt.hist(
    iter_map,
    bins=bins,
    density=True,
    histtype="step",
    linewidth=2,
    label="Iterative ensemble mean"
)

plt.xlabel("Pixel value [K]")
plt.ylabel("Probability density")
plt.title("Pixel-value distribution comparison")
plt.xlim(xmin, xmax)
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
import glob
import re
import numpy as np
import healpy as hp

# -------------------------
# Files
# -------------------------
mask_file = "mask_64.fits"

cmbpaint_files = sorted(glob.glob("inpainted_maps/Inpainted_Maps/Inpainted_T_*.fits"))[:100]
iter_files = sorted(glob.glob("iterative_inpainted_maps/*.fits"))[:100]

# -------------------------
# Load mask
# -------------------------
mask = hp.read_map(mask_file, field=0, verbose=False)
nside = hp.get_nside(mask)

obs = mask > 0.5
miss = ~obs

# -------------------------
# Boundary jump statistic
# -------------------------
def boundary_jump(m, obs, miss, nside):
    jumps = []

    for pix in np.where(miss)[0]:
        neigh = hp.get_all_neighbours(nside, pix)
        neigh = neigh[neigh >= 0]

        obs_neigh = neigh[obs[neigh]]

        if len(obs_neigh) > 0:
            jumps.extend(m[pix] - m[obs_neigh])

    jumps = np.asarray(jumps)

    return {
        "mean_jump": np.mean(jumps),
        "rms_jump": np.sqrt(np.mean(jumps**2)),
        "mean_abs_jump": np.mean(np.abs(jumps)),
        "n_pairs": len(jumps),
    }


def ensemble_boundary_test(files, label):
    rms_list = []
    abs_list = []

    for f in files:
        m = hp.read_map(f, field=0, verbose=False)

        if hp.get_nside(m) != nside:
            m = hp.ud_grade(m, nside_out=nside)

        stat = boundary_jump(m, obs, miss, nside)

        rms_list.append(stat["rms_jump"])
        abs_list.append(stat["mean_abs_jump"])

    print("\n" + label)
    print("Number of maps:", len(files))
    print("Mean RMS boundary jump:", np.mean(rms_list))
    print("Std  RMS boundary jump:", np.std(rms_list))
    print("Mean |boundary jump|:", np.mean(abs_list))
    print("Std  |boundary jump|:", np.std(abs_list))


ensemble_boundary_test(cmbpaint_files, "CMB-PAInT")
ensemble_boundary_test(iter_files, "Iterative")
```
